# BCG GenAI Financial Analysis

Analysis of 2023–2025 Form 10-K data for Microsoft, Tesla, and Apple. Values are in millions of U.S. dollars.

```python
import pandas as pd
import matplotlib.pyplot as plt
pd.set_option('display.float_format', lambda x: f'{x:,.2f}')
```

```python
data = [
['Microsoft',2023,211915,72361,411976,205753,87582],
['Microsoft',2024,245122,88136,512163,243686,118548],
['Microsoft',2025,281724,101832,619858,275524,136162],
['Tesla',2023,96773,14997,106618,43009,13256],
['Tesla',2024,97690,7091,122070,48390,14923],
['Tesla',2025,94827,5479,137806,54941,14647],
['Apple',2023,383285,96995,352583,290437,110543],
['Apple',2024,391035,93736,364980,308030,118254],
['Apple',2025,416161,112010,359241,285507,111482]
]
df = pd.DataFrame(data, columns=['Company','Year','Total Revenue','Net Income','Total Assets','Total Liabilities','Operating Cash Flow'])
df
```

```python
df = df.sort_values(['Company','Year']).reset_index(drop=True)
for metric in ['Total Revenue','Net Income','Total Assets','Total Liabilities','Operating Cash Flow']:
    df[f'{metric} Growth (%)'] = df.groupby('Company')[metric].pct_change() * 100
df['Net Margin (%)'] = df['Net Income'] / df['Total Revenue'] * 100
df['Liabilities-to-Assets (%)'] = df['Total Liabilities'] / df['Total Assets'] * 100
df.round(2)
```

```python
fig, axes = plt.subplots(1,2,figsize=(14,5))
for company, group in df.groupby('Company'):
    axes[0].plot(group['Year'], group['Total Revenue'], marker='o', label=company)
    axes[1].plot(group['Year'], group['Net Margin (%)'], marker='o', label=company)
axes[0].set_title('Total Revenue'); axes[0].set_ylabel('USD millions')
axes[1].set_title('Net Margin'); axes[1].set_ylabel('Percent')
for ax in axes: ax.set_xlabel('Year'); ax.legend(); ax.grid(alpha=.3)
plt.tight_layout(); plt.show()
```

## Findings

Microsoft shows the strongest expansion: revenue, net income, and operating cash flow rise substantially through 2025. Apple remains highly profitable and produces strong operating cash flow; its 2025 liabilities decline from 2024. Tesla's revenue is comparatively flat and net income declines after 2023, so the chatbot should distinguish balance-sheet growth from improving profitability.

The cleaned dataset supports chatbot questions about revenue growth, profitability, cash generation, and leverage.
